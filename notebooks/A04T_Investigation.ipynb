```python
import mne
import numpy as np
```

```python
file_path = "../data/BCI_IV_2a/A04T.gdf"

raw = mne.io.read_raw_gdf(
    file_path,
    preload=True
)

print(raw)

events, event_dict = mne.events_from_annotations(raw)

print("\nEvent dictionary:")
print(event_dict)

print("\nEvent counts:")

unique, counts = np.unique(events[:, 2], return_counts=True)

for event, count in zip(unique, counts):
    print(event, count)
```
